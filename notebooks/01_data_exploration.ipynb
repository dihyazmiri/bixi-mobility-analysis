```python
# Lister les fichiers disponibles
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")

for f in sorted(RAW.glob("*.csv")):
    print(f.name, f"{f.stat().st_size / 1e6:.1f} MB")
```

```python
# Charger un échantillon pour explorer les types de colonnes
sample = pd.read_csv(RAW / "DonneesOuvertes.csv", nrows=1000)
print(sample.dtypes)
sample.head(10)
```

```python
# Vérifier les timestamps bruts et les valeurs manquantes
sample[["STARTTIMEMS", "ENDTIMEMS"]].head(10)
```

```python
sample.isna().sum()
```

```python
# Convertir les timestamps Unix en dates lisibles
sample["start_time"] = pd.to_datetime(sample["STARTTIMEMS"], unit="ms")
sample["end_time"] = pd.to_datetime(sample["ENDTIMEMS"], unit="ms")

sample[["STARTTIMEMS", "start_time", "ENDTIMEMS", "end_time"]].head(10)
```

```python
# Calculer la durée des trajets et repérer les valeurs aberrantes
sample["duration_min"] = (sample["end_time"] - sample["start_time"]).dt.total_seconds() / 60
sample["duration_min"].describe()
```

```python
dtype_map = {
    "STARTSTATIONNAME": "category",
    "STARTSTATIONARRONDISSEMENT": "category",
    "STARTSTATIONLATITUDE": "float32",
    "STARTSTATIONLONGITUDE": "float32",
    "ENDSTATIONNAME": "category",
    "ENDSTATIONARRONDISSEMENT": "category",
    "ENDSTATIONLATITUDE": "float32",
    "ENDSTATIONLONGITUDE": "float32",
}

trips = pd.read_csv(
    RAW / "DonneesOuvertes.csv",
    dtype=dtype_map
)

print(f"Shape: {trips.shape}")
trips.info(memory_usage="deep")
```

```python
# Convertir les timestamps sur tout le dataset
trips["start_time"] = pd.to_datetime(trips["STARTTIMEMS"], unit="ms")
trips["end_time"] = pd.to_datetime(trips["ENDTIMEMS"], unit="ms")

print(trips["start_time"].min(), "→", trips["start_time"].max())
```

```python
trips.isna().sum()
```

```python
trips["duration_min"] = (trips["end_time"] - trips["start_time"]).dt.total_seconds() / 60
trips["duration_min"].describe()
```

```python
# Combien de trajets sont extrêmes ?
print("< 1 minute:", (trips["duration_min"] < 1).sum())
print("> 120 minutes (2h):", (trips["duration_min"] > 120).sum())
print("> 1440 minutes (24h):", (trips["duration_min"] > 1440).sum())

# Percentiles hauts pour voir où se situe la "queue" de la distribution
trips["duration_min"].quantile([0.95, 0.99, 0.999, 0.9999])
```

```python
trips["start_time"].dt.to_period("M").value_counts().sort_index()
```

```python
# Étape 1 — Séparer les trajets incomplets (pas d'heure d'arrivée) — signal métier à part
trips_incomplete = trips[trips["end_time"].isna()].copy()

print(f"Trajets incomplets (vélo non retourné): {len(trips_incomplete):,}")
```

```python
# Étape 2 — Ne garder que les trajets complets pour la suite du nettoyage
trips_complete = trips[trips["end_time"].notna()].copy()

print(f"Trajets complets: {len(trips_complete):,}")
```

```python
# Étape 3 — Sur les trajets complets uniquement, appliquer le filtre de durée
trips_clean = trips_complete[
    (trips_complete["duration_min"] >= 1) &
    (trips_complete["duration_min"] <= 120)
].copy()

print(f"Avant filtre durée: {len(trips_complete):,} trajets")
print(f"Après filtre durée: {len(trips_clean):,} trajets")
print(f"Trajets retirés (durée aberrante): {len(trips_complete) - len(trips_clean):,} ({(1 - len(trips_clean)/len(trips_complete))*100:.2f}%)")
```

```python
# Vérifier les bornes géographiques
print("Latitude départ:", trips_clean["STARTSTATIONLATITUDE"].min(), "→", trips_clean["STARTSTATIONLATITUDE"].max())
print("Longitude départ:", trips_clean["STARTSTATIONLONGITUDE"].min(), "→", trips_clean["STARTSTATIONLONGITUDE"].max())
```

```python
# Combien de lignes ont des coordonnées à 0 (départ ou arrivée) ?
zero_coords = (
    (trips_clean["STARTSTATIONLATITUDE"] == 0) |
    (trips_clean["STARTSTATIONLONGITUDE"] == 0) |
    (trips_clean["ENDSTATIONLATITUDE"] == 0) |
    (trips_clean["ENDSTATIONLONGITUDE"] == 0)
)

print(f"Lignes avec coordonnées à 0: {zero_coords.sum():,} ({zero_coords.sum()/len(trips_clean)*100:.3f}%)")
```

```python
# Retirer les lignes avec coordonnées invalides
trips_clean = trips_clean[~zero_coords].copy()

print(f"Trajets après filtre géographique: {len(trips_clean):,}")
```

```python
print("Latitude départ:", trips_clean["STARTSTATIONLATITUDE"].min(), "→", trips_clean["STARTSTATIONLATITUDE"].max())
print("Longitude départ:", trips_clean["STARTSTATIONLONGITUDE"].min(), "→", trips_clean["STARTSTATIONLONGITUDE"].max())
print("Latitude arrivée:", trips_clean["ENDSTATIONLATITUDE"].min(), "→", trips_clean["ENDSTATIONLATITUDE"].max())
print("Longitude arrivée:", trips_clean["ENDSTATIONLONGITUDE"].min(), "→", trips_clean["ENDSTATIONLONGITUDE"].max())
```

```python
PROCESSED = Path("../data/processed")

trips_clean.to_parquet(PROCESSED / "trips_clean_2024.parquet", index=False)
trips_incomplete.to_parquet(PROCESSED / "trips_incomplete_2024.parquet", index=False)

print("Fichiers sauvegardés:")
for f in sorted(PROCESSED.glob("*.parquet")):
    print(f.name, f"{f.stat().st_size / 1e6:.1f} MB")
```
